```python
def process_sampleset(sampleset, stockcodes):
    """Read in sampleset returned from sample_cqm command and display solution."""

    # Find the first feasible solution
    first_run = True
    feasible = False
    for sample, feas in sampleset.data(fields=['sample','is_feasible']):
        if first_run:
            best_sample = sample
        if feas:
            best_sample = sample
            feasible = True
            break

    # Print the solution as which stocks to buy
    print("Solution:\n")
    if not feasible:
        print("No feasible solution found.\n")
    else:
        print("Best feasible solution found:")
        for stk in stockcodes:
            if best_sample[f's_{stk}'] == 1:
                print(stk)
    print("\n")
```

## 제약조건을 "더 강하게" 만드는 방법

CQM(Constrained Quadratic Model)에서 제약조건은 이미 **하드 제약(hard constraint)**입니다. 즉, 제약조건을 위반하는 해는 절대 허용되지 않습니다.

하지만 샘플러가 제약조건을 만족하는 해를 찾지 못하는 경우가 있습니다. 이는 다음 이유 때문일 수 있습니다:

1. **샘플링 시간이 부족**: 더 긴 시간을 주면 더 나은 해를 찾을 수 있습니다.
2. **문제가 해결 불가능**: 예산이 너무 낮아서 제약조건을 만족하는 해가 없을 수 있습니다.
3. **샘플러의 한계**: 양자 하이브리드 샘플러도 완벽하지 않을 수 있습니다.

### 해결 방법:

1. **`time_limit` 파라미터 증가**: `opt.sample_cqm(time_limit=20.0)` 같이 시간을 늘립니다.
2. **예산 조정**: 예산이 너무 낮으면 해가 없을 수 있습니다.
3. **제약조건 재검토**: 제약조건이 논리적으로 올바른지 확인합니다.

**참고**: CQM에서는 제약조건의 "강도"를 조절할 수 없습니다. 제약조건은 만족되거나 위반되거나 둘 중 하나입니다.
